```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:root123@localhost/ecommerce_analytics"
)

conn = engine.connect()
print("✅ Connected Successfully!")
conn.close()
```

```python
import pandas as pd

# Read the master dataset
master = pd.read_csv("../data_cleaned/master_ecommerce.csv")

# Import into MySQL
master.to_sql(
    name="products",
    con=engine,
    if_exists="replace",
    index=False
)

print("✅ Data Imported Successfully!")
```
