```python
import pandas as pd
```

```python
amazon = pd.read_csv(r"C:\project\CartIQ\data_raw\amazon.csv")
flipkart = pd.read_csv(r"C:\project\CartIQ\data_raw\flipkart.csv")
myntra=pd.read_csv(r"C:\project\CartIQ\data_raw\Myntra.csv")
```

```python
print("Amazon")
print(amazon.info())
print(amazon.isnull().sum())
print("Duplicates:", amazon.duplicated().sum())

print("\nFlipkart")
print(flipkart.info())
print(flipkart.isnull().sum())
print("Duplicates:", flipkart.duplicated().sum())

print("\nMyntra")
print(myntra.info())
print(myntra.isnull().sum())
print("Duplicates:", myntra.duplicated().sum())
```

```python
amazon.drop_duplicates(inplace=True)
flipkart.drop_duplicates(inplace=True)
myntra.drop_duplicates(inplace=True)
```

```python
amazon["rating_count"] = amazon["rating_count"].fillna(0)

myntra["ratingCount"] = myntra["ratingCount"].fillna(0)
myntra["avg_rating"] = myntra["avg_rating"].fillna(0)


```

```python
print(flipkart.columns.tolist())
print(flipkart.columns)
```

```python
flipkart["product_rating"] = flipkart["product_rating"].fillna(0)
flipkart["overall_rating"] = flipkart["overall_rating"].fillna(0)
flipkart["brand"] = flipkart["brand"].fillna("Unknown")
```

```python
flipkart.rename(columns={
    "uniq_id": "product_id",
    "product_category_tree": "category",
    "retail_price": "actual_price",
    "description": "about_product",
    "product_rating": "rating"
}, inplace=True)
```

```python
myntra.rename(columns={
    "p_id": "product_id",
    "name": "product_name",
    "products": "category",
    "avg_rating": "rating",
    "ratingCount": "rating_count",
    "description": "about_product"
}, inplace=True)
```

```python
amazon["source"] = "Amazon"

flipkart["source"] = "Flipkart"

myntra["source"] = "Myntra"
```

```python
print(amazon.columns.tolist())

print(flipkart.columns.tolist())

print(myntra.columns.tolist())
```

```python
amazon.to_csv("C:\\project\\CartIQ\\data_cleaned\\amazon_clean.csv", index=False)

flipkart.to_csv("C:\\project\\CartIQ\\data_cleaned\\flipkart_clean.csv", index=False)

myntra.to_csv("C:\\project\\CartIQ\\data_cleaned\\myntra_clean.csv", index=False)
```
