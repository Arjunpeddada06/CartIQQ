```python
import pandas as pd
```

```python
amazon=pd.read_csv("../data_cleaned/amazon_clean.csv")
flipkart=pd.read_csv("../data_cleaned/flipkart_clean.csv")
myntra=pd.read_csv("../data_cleaned/myntra_clean.csv")
```

```python
print(amazon.columns.tolist())
print()
print(flipkart.columns.tolist())
print()
print(myntra.columns.tolist())
```

```python
amazon = amazon.rename(columns={
    "discounted_price": "price",
    "img_link": "image"
})

amazon = amazon[
    [
        "product_id",
        "product_name",
        "category",
        "price",
        "rating",
        "rating_count",
        "about_product",
        "image",
        "source"
    ]
]

amazon["brand"] = "Unknown"
```

```python
flipkart = flipkart.rename(columns={
    "discounted_price": "price",
    "image": "image"
})

flipkart = flipkart[
    [
        "product_id",
        "product_name",
        "category",
        "brand",
        "price",
        "rating",
        "about_product",
        "image",
        "source"
    ]
]

flipkart["rating_count"] = 0
```

```python
myntra = myntra.rename(columns={
    "price": "price",
    "img": "image"
})

myntra = myntra[
    [
        "product_id",
        "product_name",
        "category",
        "brand",
        "price",
        "rating",
        "rating_count",
        "about_product",
        "image",
        "source"
    ]
]
```

Merging all datasets

```python
master=pd.concat([amazon,flipkart,myntra],ignore_index=True)
```

```python
print(master.shape)
print()
print(master.columns)
print()
print(master.head())
```
